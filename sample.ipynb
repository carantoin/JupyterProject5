```python
import numpy as np

# 1
arr1 = np.random.randint(99, 101, size=5)
print("1. Случайные числа:", arr1)


# 2
arr2 = np.full(10, 7)
print("2. Массив из семёрок:", arr2)


# 3
arr3 = np.arange(0, 1.1, 0.1)
print("3. Массив от 0 до 1:", arr3)


# 4
arr4 = np.array([10, 20, 30, 40, 50, 60, 70])
sum_even_pos = arr4[::2].sum()
print("4. Массив:", arr4)
print("   Сумма элементов на чётных позициях:", sum_even_pos)


# 5
matrix5 = np.array([[1, 2, 3],
                    [0, 1, 4],
                    [5, 6, 0]], dtype=float)

det = np.linalg.det(matrix5)
print("5. Определитель:", round(det, 4))

if det != 0:
    inverse_matrix = np.linalg.inv(matrix5)
    print("   Обратная матрица:\n", inverse_matrix)
else:
    print("   Матрица вырожденная, обратной не существует.")


# 6
matrix6 = np.arange(1, 17).reshape(4, 4)
print("6. Матрица 4×4:\n", matrix6)

main_diag = np.diag(matrix6)
anti_diag = np.diag(np.fliplr(matrix6))
print("   Главная диагональ:", main_diag)
print("   Побочная диагональ:", anti_diag)


# 7
matrix7 = np.random.randint(1, 10, size=(5, 5))
print("7. Матрица 5×5:\n", matrix7)

max_val = matrix7.max()
indices = np.argwhere(matrix7 == max_val)
print(f"   Максимум = {max_val}")
print("   Индексы максимума (строки, столбцы):\n", indices)


# 8
a = np.array([1, 2, 3, 4, 5])
b = np.array([2, 4, 6, 8, 10])

cos_sim = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
print("8. Массив a:", a)
print("   Массив b:", b)
print("   Косинусное сходство:", cos_sim)


# 9
arr9 = np.array([5, 2, 8, 1, 9, 3])
asc  = np.sort(arr9)              # возрастание
desc = np.sort(arr9)[::-1]        # убывание (переворот)
print("9. Исходный массив:", arr9)
print("   По возрастанию:", asc)
print("   По убыванию:  ", desc)


# 10
labels = np.array([0, 2, 1, 3])
num_classes = labels.max() + 1  # 4 класса (0, 1, 2, 3)

one_hot = np.eye(num_classes, dtype=int)[labels]

print("10. Метки:", labels)
print("    One-hot кодирование:\n", one_hot)
```
